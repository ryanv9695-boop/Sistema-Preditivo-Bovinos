```python
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit
import os

# 1. Carregamento dos dados com o caminho atualizado
caminho_csv = '/home/vicom03/CID/data/raw/A Dataset on Nelore Cattle Supplementation in the Midwest Region of Brazil for Integrated Analysis of Supplementation, Satellite Remote Sensing, and Weather Variables/data/Complete_Dataset.csv'
df = pd.read_csv(caminho_csv)

# Exibe as colunas para conferirmos o nome exato do ID do animal
print("Colunas do dataset:", df.columns.tolist())

# 2. Definição da coluna de agrupamento
# ATENÇÃO: Se na lista acima o nome for diferente (ex: 'ID', 'ANIMAL_ID'), altere aqui!
coluna_grupo = 'ANIMAL' 

# 3. Primeira divisão: 80% Treino e 20% Temporário
gss_train_temp = GroupShuffleSplit(n_splits=1, train_size=0.8, random_state=42)
train_idx, temp_idx = next(gss_train_temp.split(df, groups=df[coluna_grupo]))

df_train = df.iloc[train_idx]
df_temp = df.iloc[temp_idx]

# 4. Segunda divisão: divide os 20% restantes pela metade (10% Validação, 10% Teste)
gss_val_test = GroupShuffleSplit(n_splits=1, train_size=0.5, random_state=42)
val_idx, test_idx = next(gss_val_test.split(df_temp, groups=df_temp[coluna_grupo]))

df_val = df_temp.iloc[val_idx]
df_test = df_temp.iloc[test_idx]

# 5. Verificação das divisões
print(f"\n--- Resumo da Divisão ---")
print(f"Total de registros: {len(df)}")
print(f"Treino (80%): {len(df_train)} registros | Animais únicos: {df_train[coluna_grupo].nunique()}")
print(f"Validação (10%): {len(df_val)} registros | Animais únicos: {df_val[coluna_grupo].nunique()}")
print(f"Teste (10%): {len(df_test)} registros | Animais únicos: {df_test[coluna_grupo].nunique()}")

# 6. Salvando os arquivos processados
os.makedirs('../data/processed', exist_ok=True)
df_train.to_csv('../data/processed/treino.csv', index=False)
df_val.to_csv('../data/processed/validacao.csv', index=False)
df_test.to_csv('../data/processed/teste.csv', index=False)
print("\nArquivos salvos com sucesso na pasta data/processed/")
```
