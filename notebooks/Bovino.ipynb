```python
import pandas as pd

# O '../' faz o Python voltar para a pasta principal antes de entrar em data/processed
caminho_csv = '../data/processed/treino.csv'
df = pd.read_csv(caminho_csv)

# Mostra as 5 primeiras linhas para confirmar
df.head()
```

```python
# Lista o nome de todas as colunas do dataset de treino
print(df.columns.tolist())
```

```python
# Definindo a variável alvo (y) e as preditoras (X)
y = df['ADG']
colunas_para_remover = ['ADG', 'FINAL_WEIGHT', 'START_DATE', 'FINAL_DATE', 'ANIMAL']
X = df.drop(columns=colunas_para_remover)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

X_encoded = pd.get_dummies(X)
modelo_rf = RandomForestRegressor(n_estimators=100, random_state=42)

print("Treinando o modelo. Isso pode levar alguns segundos...")
modelo_rf.fit(X_encoded, y)

previsoes = modelo_rf.predict(X_encoded)
erro_medio = mean_absolute_error(y, previsoes)
print(f"Erro Médio Absoluto no Treino: {erro_medio:.4f} kg/dia")
```

```python
import pandas as pd

# O parâmetro decimal=',' converte todas as vírgulas para pontos automaticamente
caminho_csv = '../data/processed/treino.csv'
df = pd.read_csv(caminho_csv, decimal=',')

df.head()
```

```python
# Definindo a variável alvo (y) e as preditoras (X)
y = df['ADG']
colunas_para_remover = ['ADG', 'FINAL_WEIGHT', 'START_DATE', 'FINAL_DATE', 'ANIMAL']
X = df.drop(columns=colunas_para_remover)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

X_encoded = pd.get_dummies(X)
modelo_rf = RandomForestRegressor(n_estimators=100, random_state=42)

print("Treinando o modelo. Isso pode levar alguns segundos...")
modelo_rf.fit(X_encoded, y)

previsoes = modelo_rf.predict(X_encoded)
erro_medio = mean_absolute_error(y, previsoes)
print(f"Erro Médio Absoluto no Treino: {erro_medio:.4f} kg/dia")
```

```python
# 1. Carrega os dados de validação (lembrando da vírgula decimal!)
df_val = pd.read_csv('../data/processed/validacao.csv', decimal=',')

# 2. Separa X e y da validação
y_val = df_val['ADG']
X_val = df_val.drop(columns=colunas_para_remover)

# 3. Converte os textos em números igual fizemos no treino
X_val_encoded = pd.get_dummies(X_val)

# 4. Alinha as colunas (Garante que a validação tenha as mesmas colunas do treino)
# Isso é essencial caso falte alguma categoria no arquivo de validação
X_val_encoded = X_val_encoded.reindex(columns=X_encoded.columns, fill_value=0)

# 5. A Prova Real: Pede para o modelo prever usando os dados novos
previsoes_val = modelo_rf.predict(X_val_encoded)

# 6. Calcula o erro nesses dados novos
erro_val = mean_absolute_error(y_val, previsoes_val)
print(f"Erro Médio Absoluto na Validação (A Prova Real): {erro_val:.4f} kg/dia")
```
